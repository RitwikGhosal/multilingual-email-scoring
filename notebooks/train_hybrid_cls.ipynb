```python
CEFR_TO_ID = {
    "A1": 0,
    "A2": 1,
    "B1": 2,
    "B2": 3,
    "C1": 4,
    "C2": 5,
}

ID_TO_CEFR = {
    value: key
    for key, value in CEFR_TO_ID.items()
}


def prepare_labels(frame):
    frame = frame.copy()

    frame["grammar_labels"] = (
        pd.to_numeric(
            frame["grammar"],
            errors="coerce",
        )
        .fillna(-100)
        .astype(float)
    )

    frame["content_labels"] = (
        pd.to_numeric(
            frame["content"],
            errors="coerce",
        )
        .fillna(-100)
        .astype(float)
    )

    # Exact CEFR label, used for evaluation.
    frame["cefr_labels"] = (
        frame["cefr"]
        .map(CEFR_TO_ID)
        .fillna(-100)
        .astype(int)
    )

    # Exact or interval endpoints, used for training.
    frame["cefr_min_labels"] = (
        pd.to_numeric(
            frame["cefr_min"],
            errors="coerce",
        )
        .fillna(-100)
        .astype(int)
    )

    frame["cefr_max_labels"] = (
        pd.to_numeric(
            frame["cefr_max"],
            errors="coerce",
        )
        .fillna(-100)
        .astype(int)
    )

    return frame


train_df = prepare_labels(train_df)
validation_df = prepare_labels(
    validation_df
)

print(
    "Exact CEFR training rows:",
    (train_df["cefr_labels"] != -100).sum(),
)

print(
    "All CEFR-supervised training rows:",
    (train_df["cefr_min_labels"] != -100).sum(),
)

print(
    "Interval CEFR training rows:",
    (
        (
            train_df["cefr_min_labels"] != -100
        )
        &
        (
            train_df["cefr_min_labels"]
            != train_df["cefr_max_labels"]
        )
    ).sum(),
)
```

```python
class MultitaskDataCollator:

    def __init__(
        self,
        tokenizer,
        pad_to_multiple_of=8,
    ):
        self.tokenizer = tokenizer
        self.pad_to_multiple_of = pad_to_multiple_of

    def __call__(self, features):
        model_features = [
            {
                key: feature[key]
                for key in self.tokenizer.model_input_names
                if key in feature
            }
            for feature in features
        ]

        batch = self.tokenizer.pad(
            model_features,
            padding=True,
            pad_to_multiple_of=self.pad_to_multiple_of,
            return_tensors="pt",
        )

        for label_name in [
            "grammar_labels",
            "content_labels",
        ]:
            batch[label_name] = torch.tensor(
                [
                    feature[label_name]
                    for feature in features
                ],
                dtype=torch.float32,
            )

        for label_name in [
            "cefr_labels",
            "cefr_min_labels",
            "cefr_max_labels",
        ]:
            batch[label_name] = torch.tensor(
                [
                    feature[label_name]
                    for feature in features
                ],
                dtype=torch.long,
            )

        return batch

data_collator = MultitaskDataCollator(
    tokenizer=tokenizer,
    pad_to_multiple_of=8,
)
```

```python
import math
import torch
import torch.nn as nn
import torch.nn.functional as F

from transformers import AutoModel
from peft import (
    LoraConfig,
    TaskType,
    get_peft_model,
)
```

```python
class OrderedCEFRHead(nn.Module):
    """
    Produces five ordered logits:

    P(level > A1)
    P(level > A2)
    P(level > B1)
    P(level > B2)
    P(level > C1)
    """

    def __init__(self, input_size):
        super().__init__()

        self.score = nn.Linear(
            input_size,
            1,
        )

        # Initial thresholds:
        # [-2, -1, 0, 1, 2]
        self.base_threshold = nn.Parameter(
            torch.tensor(-2.0)
        )

        inverse_softplus_one = math.log(
            math.exp(1.0) - 1.0
        )

        self.raw_steps = nn.Parameter(
            torch.full(
                (4,),
                inverse_softplus_one,
            )
        )

    def get_thresholds(self):
        positive_steps = F.softplus(
            self.raw_steps
        )

        remaining_thresholds = (
            self.base_threshold
            + torch.cumsum(
                positive_steps,
                dim = 0,
            )
        )

        return torch.cat([
            self.base_threshold.reshape(1),
            remaining_thresholds,
        ])

    def forward(self, hidden_states):
        score = self.score(hidden_states)
        thresholds = self.get_thresholds()
        return score - thresholds.unsqueeze(0)
```

```python
class XLMRMultitaskModel(nn.Module):

    def __init__(
        self,
        model_name="xlm-roberta-base",
        projection_size=256,
        dropout=0.10,
    ):
        super().__init__()

        base_encoder = AutoModel.from_pretrained(
            model_name
        )

        lora_config = LoraConfig(
            task_type=TaskType.FEATURE_EXTRACTION,
            r=8,
            lora_alpha=16,
            lora_dropout=0.10,
            target_modules=[
                "query",
                "value",
            ],
            bias="none",
        )

        self.encoder = get_peft_model(
            base_encoder,
            lora_config,
        )

        hidden_size = (
            base_encoder.config.hidden_size
        )

        self.shared_projection = nn.Sequential(
            nn.LayerNorm(hidden_size),
            nn.Linear(
                hidden_size,
                projection_size,
            ),
            nn.GELU(),
            nn.Dropout(dropout),
        )

        self.grammar_head = nn.Linear(
            projection_size,
            1,
        )

        self.content_head = nn.Linear(
            projection_size,
            1,
        )

        self.cefr_head = OrderedCEFRHead(
            projection_size
        )

    def forward(
        self,
        input_ids,
        attention_mask,
        grammar_labels=None,
        content_labels=None,
        cefr_labels=None,
        cefr_min_labels=None,
        cefr_max_labels=None,
    ):
        encoder_output = self.encoder(
            input_ids=input_ids,
            attention_mask=attention_mask,
        )

        # Representation of XLM-R's first <s> token.
        pooled_output = (
            encoder_output.last_hidden_state[:, 0]
        )

        features = self.shared_projection(
            pooled_output
        )

        grammar_predictions = (
            torch.sigmoid(
                self.grammar_head(features)
            )
            .squeeze(-1)
            * 5.0
        )

        content_predictions = (
            torch.sigmoid(
                self.content_head(features)
            )
            .squeeze(-1)
            * 4.0
        )

        cefr_logits = self.cefr_head(
            features
        )

        # Number of ordinal thresholds passed:
        # 0=A1, ..., 5=C2.
        cefr_predictions = (
            torch.sigmoid(cefr_logits)
            > 0.5
        ).sum(dim=1)

        output = {
            "grammar_predictions":
                grammar_predictions,

            "content_predictions":
                content_predictions,

            "cefr_logits":
                cefr_logits,

            "cefr_predictions":
                cefr_predictions,
        }

        losses = []

        # -------------------------------------------------
        # Grammar loss
        # -------------------------------------------------
        if grammar_labels is not None:
            grammar_mask = (
                grammar_labels != -100
            )

            if grammar_mask.any():
                grammar_loss = F.mse_loss(
                    (
                        grammar_predictions[
                            grammar_mask
                        ]
                        / 5.0
                    ),
                    (
                        grammar_labels[
                            grammar_mask
                        ]
                        / 5.0
                    ),
                )

                output["grammar_loss"] = (
                    grammar_loss
                )

                losses.append(
                    grammar_loss
                )

        # -------------------------------------------------
        # Content loss
        # -------------------------------------------------
        if content_labels is not None:
            content_mask = (
                content_labels != -100
            )

            if content_mask.any():
                content_loss = F.mse_loss(
                    (
                        content_predictions[
                            content_mask
                        ]
                        / 4.0
                    ),
                    (
                        content_labels[
                            content_mask
                        ]
                        / 4.0
                    ),
                )

                output["content_loss"] = (
                    content_loss
                )

                losses.append(
                    content_loss
                )

        # -------------------------------------------------
        # Interval-aware ordinal CEFR loss
        # -------------------------------------------------
        if (
            cefr_min_labels is not None
            and cefr_max_labels is not None
        ):
            cefr_mask = (
                (cefr_min_labels >= 0)
                &
                (cefr_max_labels >= 0)
            )

            if cefr_mask.any():
                valid_logits = (
                    cefr_logits[cefr_mask]
                )

                minimum_levels = (
                    cefr_min_labels[
                        cefr_mask
                    ]
                )

                maximum_levels = (
                    cefr_max_labels[
                        cefr_mask
                    ]
                )

                thresholds = torch.arange(
                    5,
                    device=cefr_logits.device,
                ).unsqueeze(0)

                # Thresholds that every possible
                # label in the interval passes.
                known_positive = (
                    thresholds
                    < minimum_levels.unsqueeze(1)
                )

                # Thresholds that no possible
                # label in the interval passes.
                known_negative = (
                    thresholds
                    >= maximum_levels.unsqueeze(1)
                )

                known_thresholds = (
                    known_positive
                    | known_negative
                )

                ordinal_targets = (
                    known_positive.float()
                )

                cefr_loss_matrix = (
                    F.binary_cross_entropy_with_logits(
                        valid_logits,
                        ordinal_targets,
                        reduction="none",
                    )
                )

                cefr_loss = (
                    cefr_loss_matrix[
                        known_thresholds
                    ].mean()
                )

                output["cefr_loss"] = (
                    cefr_loss
                )

                losses.append(
                    cefr_loss
                )

        if losses:
            output["loss"] = torch.stack(
                losses
            ).sum()

        return output
```

```python
from torch.utils.data import DataLoader

TRAIN_BATCH_SIZE = 8
VALIDATION_BATCH_SIZE = 16
GRADIENT_ACCUMULATION_STEPS = 2

train_loader = DataLoader(
    tokenized_datasets["train"],
    batch_size=TRAIN_BATCH_SIZE,
    shuffle=True,
    collate_fn=data_collator,
    pin_memory=True,
    num_workers=2,
)

validation_loader = DataLoader(
    tokenized_datasets["validation"],
    batch_size=VALIDATION_BATCH_SIZE,
    shuffle=False,
    collate_fn=data_collator,
    pin_memory=True,
    num_workers=2,
)

print("Training batches:", len(train_loader))
print("Validation batches:", len(validation_loader))
print(
    "Effective batch size:",
    TRAIN_BATCH_SIZE
    * GRADIENT_ACCUMULATION_STEPS,
)
```

```python
from scipy.stats import pearsonr
from sklearn.metrics import accuracy_score

import numpy as np
import torch


def safe_pearson(targets, predictions):
    targets = np.asarray(targets)
    predictions = np.asarray(predictions)

    if (
        len(targets) < 2
        or np.std(targets) == 0
        or np.std(predictions) == 0
    ):
        return 0.0

    return float(
        pearsonr(
            targets,
            predictions,
        ).statistic
    )


@torch.no_grad()
def evaluate_model(
    model,
    data_loader,
    device,
):
    model.eval()

    losses = []

    grammar_targets = []
    grammar_predictions = []

    content_targets = []
    content_predictions = []

    cefr_targets = []
    cefr_hard_predictions = []
    cefr_soft_predictions = []

    for batch in data_loader:
        batch = {
            key: value.to(
                device,
                non_blocking=True,
            )
            for key, value in batch.items()
        }

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=device.type == "cuda",
        ):
            output = model(**batch)

        if "loss" in output:
            losses.append(
                output["loss"]
                .detach()
                .float()
                .item()
            )

        # -------------------------------------------------
        # Grammar metrics: only rows with grammar labels
        # -------------------------------------------------
        grammar_mask = (
            batch["grammar_labels"] != -100
        )

        if grammar_mask.any():
            grammar_targets.extend(
                batch["grammar_labels"][grammar_mask]
                .detach()
                .float()
                .cpu()
                .numpy()
            )

            grammar_predictions.extend(
                output["grammar_predictions"][grammar_mask]
                .detach()
                .float()
                .cpu()
                .numpy()
            )

        # -------------------------------------------------
        # Content metrics: only rows with content labels
        # -------------------------------------------------
        content_mask = (
            batch["content_labels"] != -100
        )

        if content_mask.any():
            content_targets.extend(
                batch["content_labels"][content_mask]
                .detach()
                .float()
                .cpu()
                .numpy()
            )

            content_predictions.extend(
                output["content_predictions"][content_mask]
                .detach()
                .float()
                .cpu()
                .numpy()
            )

        # -------------------------------------------------
        # CEFR metrics: only rows with exact CEFR labels
        #
        # Interval-labelled examples have cefr_labels=-100,
        # so they help training/loss but are not used for
        # exact-label accuracy or Pearson correlation.
        # -------------------------------------------------
        cefr_mask = (
            batch["cefr_labels"] != -100
        )

        if cefr_mask.any():
            valid_cefr_targets = (
                batch["cefr_labels"][cefr_mask]
            )

            valid_cefr_logits = (
                output["cefr_logits"][cefr_mask]
            )

            probabilities = torch.sigmoid(
                valid_cefr_logits
            )

            hard_predictions = (
                probabilities > 0.5
            ).sum(dim=1)

            # Continuous ordinal score in [0, 5].
            # Used only for diagnostic Pearson correlation.
            soft_predictions = probabilities.sum(
                dim=1
            )

            cefr_targets.extend(
                valid_cefr_targets
                .detach()
                .cpu()
                .numpy()
            )

            cefr_hard_predictions.extend(
                hard_predictions
                .detach()
                .cpu()
                .numpy()
            )

            cefr_soft_predictions.extend(
                soft_predictions
                .detach()
                .float()
                .cpu()
                .numpy()
            )

    metrics = {
        "validation_loss": (
            float(np.mean(losses))
            if losses
            else float("nan")
        ),

        "grammar_pearson": safe_pearson(
            grammar_targets,
            grammar_predictions,
        ),

        "content_pearson": safe_pearson(
            content_targets,
            content_predictions,
        ),

        "cefr_pearson_hard": safe_pearson(
            cefr_targets,
            cefr_hard_predictions,
        ),

        "cefr_pearson_soft": safe_pearson(
            cefr_targets,
            cefr_soft_predictions,
        ),

        "cefr_accuracy": (
            float(
                accuracy_score(
                    cefr_targets,
                    cefr_hard_predictions,
                )
            )
            if cefr_targets
            else float("nan")
        ),

        # Useful sanity-check counts
        "grammar_evaluation_rows":
            len(grammar_targets),

        "content_evaluation_rows":
            len(content_targets),

        "cefr_exact_evaluation_rows":
            len(cefr_targets),
    }

    return metrics
```

```python
import random
from transformers import (
    get_linear_schedule_with_warmup,
)

SEED = 42
EPOCHS = 8

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

torch.set_float32_matmul_precision("high")

encoder_parameters = [
    parameter
    for name, parameter
    in model.named_parameters()
    if (
        parameter.requires_grad
        and name.startswith("encoder.")
    )
]

head_parameters = [
    parameter
    for name, parameter
    in model.named_parameters()
    if (
        parameter.requires_grad
        and not name.startswith("encoder.")
    )
]

optimizer = torch.optim.AdamW(
    [
        {
            "params": encoder_parameters,
            "lr": 2e-4,
            "weight_decay": 0.01,
        },
        {
            "params": head_parameters,
            "lr": 5e-4,
            "weight_decay": 0.01,
        },
    ]
)

updates_per_epoch = math.ceil(
    len(train_loader)
    / GRADIENT_ACCUMULATION_STEPS
)

total_training_steps = (
    updates_per_epoch * EPOCHS
)

warmup_steps = int(
    0.10 * total_training_steps
)

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_training_steps,
)


scaler = torch.amp.GradScaler(
    "cuda",
    enabled=device.type == "cuda",
)

print("Updates per epoch:", updates_per_epoch)
print("Total optimizer updates:", total_training_steps)
print("Warmup updates:", warmup_steps)
```

```python
from tqdm.auto import tqdm

history = []
best_score = -float("inf")
best_epoch = None
best_trainable_state = None

optimizer.zero_grad(set_to_none=True)

for epoch in range(1, EPOCHS + 1):
    model.train()

    running_loss = 0.0

    progress_bar = tqdm(
        enumerate(train_loader),
        total=len(train_loader),
        desc=f"Epoch {epoch}/{EPOCHS}",
    )

    for step, batch in progress_bar:
        batch = {
            key: value.to(
                device,
                non_blocking=True,
            )
            for key, value in batch.items()
        }

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=device.type == "cuda",
        ):
            output = model(**batch)

            loss = (
                output["loss"]
                / GRADIENT_ACCUMULATION_STEPS
            )

        scaler.scale(loss).backward()

        should_update = (
            (step + 1)
            % GRADIENT_ACCUMULATION_STEPS
            == 0
            or (step + 1) == len(train_loader)
        )

        if should_update:
            scaler.unscale_(optimizer)

            torch.nn.utils.clip_grad_norm_(
                [
                    parameter
                    for parameter
                    in model.parameters()
                    if parameter.requires_grad
                ],
                max_norm=1.0,
            )

            scaler.step(optimizer)
            scaler.update()

            scheduler.step()
            optimizer.zero_grad(
                set_to_none=True
            )

        running_loss += (
            loss.item()
            * GRADIENT_ACCUMULATION_STEPS
        )

        progress_bar.set_postfix(
            loss=(
                running_loss
                / (step + 1)
            ),
            lr=scheduler.get_last_lr()[0],
        )

    metrics = evaluate_model(
        model=model,
        data_loader=validation_loader,
        device=device,
    )

    metrics["epoch"] = epoch
    metrics["training_loss"] = (
        running_loss / len(train_loader)
    )

    # Composite checkpoint-selection score
    metrics["selection_score"] = np.mean([
        metrics["grammar_pearson"],
        metrics["content_pearson"],
        metrics["cefr_pearson_soft"],
        metrics["cefr_accuracy"],
    ])

    history.append(metrics)

    print("\nEpoch", epoch)
    for key, value in metrics.items():
        if isinstance(value, float):
            print(f"{key}: {value:.4f}")
        else:
            print(f"{key}: {value}")

    if metrics["selection_score"] > best_score:
        best_score = metrics[
            "selection_score"
        ]

        best_epoch = epoch

        # Only LoRA and prediction-head weights
        best_trainable_state = {
            name: parameter
            .detach()
            .cpu()
            .clone()

            for name, parameter
            in model.named_parameters()

            if parameter.requires_grad
        }

        print(
            "Saved best in-memory checkpoint."
        )
```

```python
@torch.no_grad()
def collect_cefr_validation_scores(
    model,
    data_loader,
    device,
):
    model.eval()

    targets = []
    soft_scores = []

    for batch in data_loader:
        batch = {
            key: value.to(device)
            for key, value in batch.items()
        }

        output = model(**batch)

        # Only exact CEFR labels are valid for calibration
        mask = batch["cefr_labels"] != -100

        if mask.any():
            logits = output["cefr_logits"][mask]

            scores = torch.sigmoid(
                logits
            ).sum(dim=1)

            targets.extend(
                batch["cefr_labels"][mask]
                .detach()
                .cpu()
                .numpy()
            )

            soft_scores.extend(
                scores.detach()
                .float()
                .cpu()
                .numpy()
            )

    return (
        np.asarray(targets),
        np.asarray(soft_scores),
    )


cefr_targets, cefr_soft_scores = (
    collect_cefr_validation_scores(
        model=model,
        data_loader=validation_loader,
        device=device,
    )
)

print("Exact calibration rows:", len(cefr_targets))
```

```python
def find_adjacent_cut(
    targets,
    scores,
    lower_level,
    upper_level,
):
    mask = np.isin(
        targets,
        [lower_level, upper_level],
    )

    pair_targets = targets[mask]
    pair_scores = scores[mask]

    unique_scores = np.unique(pair_scores)

    candidates = (
        unique_scores[:-1]
        + unique_scores[1:]
    ) / 2.0

    best_cut = None
    best_accuracy = -1.0

    for cut in candidates:
        predictions = np.where(
            pair_scores > cut,
            upper_level,
            lower_level,
        )

        accuracy = accuracy_score(
            pair_targets,
            predictions,
        )

        if accuracy > best_accuracy:
            best_accuracy = accuracy
            best_cut = float(cut)

    return best_cut, float(best_accuracy)


cut_points = []
pair_accuracies = []

for lower_level in range(5):
    upper_level = lower_level + 1

    cut, pair_accuracy = find_adjacent_cut(
        cefr_targets,
        cefr_soft_scores,
        lower_level,
        upper_level,
    )

    cut_points.append(cut)
    pair_accuracies.append(pair_accuracy)

    print(
        f"{ID_TO_CEFR[lower_level]} / "
        f"{ID_TO_CEFR[upper_level]}: "
        f"cut={cut:.4f}, "
        f"accuracy={pair_accuracy:.4f}"
    )

cut_points = np.asarray(cut_points)

assert np.all(np.diff(cut_points) > 0), (
    f"Calibration thresholds are not ordered: "
    f"{cut_points}"
)
```

```python
calibrated_predictions = np.digitize(
    cefr_soft_scores,
    bins=cut_points,
)

calibration_metrics = {
    "exact_accuracy": float(
        accuracy_score(
            cefr_targets,
            calibrated_predictions,
        )
    ),
    "pearson": safe_pearson(
        cefr_targets,
        calibrated_predictions,
    ),
    "mean_distance": float(
        np.mean(
            np.abs(
                cefr_targets
                - calibrated_predictions
            )
        )
    ),
    "within_one_level": float(
        np.mean(
            np.abs(
                cefr_targets
                - calibrated_predictions
            ) <= 1
        )
    ),
}

calibration_data = {
    "method": "adjacent_class_validation_cutpoints",
    "cut_points": cut_points.tolist(),
    "pair_accuracies": pair_accuracies,
    "class_mapping": ID_TO_CEFR,
    "exact_validation_rows": int(
        len(cefr_targets)
    ),
    "metrics": calibration_metrics,
}

with open(
    CHECKPOINT_DIR / "cefr_calibration.json",
    "w",
    encoding="utf-8",
) as file:
    json.dump(
        calibration_data,
        file,
        indent=2,
        ensure_ascii=False,
    )

print("Cut points:", cut_points)
print("Metrics:", calibration_metrics)
print(
    "Saved:",
    CHECKPOINT_DIR / "cefr_calibration.json",
)
```
